```python
df=df.drop_duplicates()
```

```python
df.shape
```

```python
df['Date']=pd.to_datetime(df['Date'])
```

```python
df.info()
```

```python
df.head()
```

```python
df['Month']=df['Date'].dt.month_name()  # month column creation
```

```python
monthly_sales=df.groupby('Month')['Total Amount'].sum()
```

```python
month_order = ["January","February","March","April","May","June","July","August","September","October","November","December"]
```

```python
monthly_sales = monthly_sales.reindex(month_order)
```

```python
df.head()
```

```python
plt.figure(figsize=(10,5))
monthly_sales.plot(marker='o')
```

```python
plt.figure(figsize=(10,5))
monthly_sales.plot(marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

```python
#Observation
#Sales changed from month to month. Some months had higher sales while others had lower sales.
```

```python
# quarter column
df['Quarter']=df['Date'].dt.quarter
```

```python
quarterly_sales=df.groupby('Quarter')['Total Amount'].sum()
```

```python
plt.figure(figsize=(8,5))

quarterly_sales.plot(marker='o')
```

```python
plt.figure(figsize=(8,5))

quarterly_sales.plot(marker='o')

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.grid(True)

plt.show()
```

```python
#observation
#Sales were different in each quarter. One quarter performed better than the others.
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"], bins=10, kde=True)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.show()
```

```python
#observation
#Most customers belong to a particular age group. This age group is the main target for the business.
```

```python
gender = df["Gender"].value_counts()

plt.figure(figsize=(6,5))

gender.plot(kind="bar")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Customers")

plt.show()
```

```python
#Observation
#Both male and female customers purchased products. One gender has slightly more customers than the other.
```

```python
top_products = df.groupby("Product Category")["Quantity"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))

sns.barplot(x=top_products.index,
            y=top_products.values)

plt.xlabel("Product Category")
plt.ylabel("Total Quantity")

plt.show()
```

```python
##Observation
#Some product categories sold more than others. These products are more popular among customers.
```

```python
revenue = df.groupby("Product Category")["Total Amount"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))

sns.barplot(x=revenue.index,
            y=revenue.values)

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")

plt.show()
```

```python
##Observation
#Some product categories generated more revenue. These products are important for business growth.
```

```python
numeric = df[["Age","Quantity","Price per Unit","Total Amount"]]

corr = numeric.corr()

plt.figure(figsize=(8,6))

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Matrix")

plt.show()
```

```python
#Observation
#Some numerical columns are related to each other. Stronger values show a better relationship between variables.
```

```python
gender_revenue = df.groupby("Gender")["Total Amount"].sum()

plt.figure(figsize=(6,5))

gender_revenue.plot(kind="bar")

plt.title("Revenue by Gender")
plt.xlabel("Gender")
plt.ylabel("Revenue")

plt.show()
```

```python
#Observation
#his chart gives extra information about customer buying behavior. It helps to understand the data better.
```
