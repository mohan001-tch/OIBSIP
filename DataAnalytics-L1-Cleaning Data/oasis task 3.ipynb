```python
numeric_columns = df.select_dtypes(include=np.number).columns
```

```python
numeric_columns
```

```python
outlier_report = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)   # 25%
    Q3 = df[col].quantile(0.75)   # 75%
    
    IQR = Q3 - Q1   # formula
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]
    
    outlier_report.append([
        col,
        Q1,
        Q3,
        lower_limit,
        upper_limit,
        len(outliers)
    ])

outlier_report = pd.DataFrame(
    outlier_report,
    columns=[
        "Column",
        "Q1",
        "Q3",
        "Lower Limit",
        "Upper Limit",
        "Outlier Count"
    ]
)

outlier_report
```

```python
for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)][col]
```

# IQR method: Used to identify outliers in numeric columns.
# Outliers found: Age, SibSp, Parch, and Fare.
# Decision: Outliers were retained as they may represent valid data.
# Reason: Removing or capping them could cause loss of useful information.

```python
df["PassengerId"] = df["PassengerId"].astype("string")
```

```python
df["Survived"] = df["Survived"].astype("int64")
df["Pclass"] = df["Pclass"].astype("int64")
df["Age"] = df["Age"].astype("float64")
df["SibSp"] = df["SibSp"].astype("int64")
df["Parch"] = df["Parch"].astype("int64")
df["Fare"] = df["Fare"].astype("float64")
```

```python
dtype_accuracy = {
    "PassengerId": "string",
    "Survived": "int64",
    "Pclass": "int64",
    "Name": "object",
    "Sex": "object",
    "Age": "float64",
    "SibSp": "int64",
    "Parch": "int64",
    "Ticket": "object",
    "Fare": "float64",
    "Cabin": "object",
    "Embarked": "object"
}

correct_dtypes = 0

for col, expected in dtype_accuracy.items():
    if str(df[col].dtype) == expected:
        correct_dtypes += 1

print("Correct data types:", correct_dtypes)
print("Total columns:", len(dtype_accuracy))
```

# PassengerId: Converted from integer to string as it is an identifier.
# Age and Fare: Stored as float values.
# Survived, Pclass, SibSp, Parch: Stored as integers.
# Categorical columns: Kept as object/string type.
